# 키보드 입력

- 키보드로 데이터를 입력받기 위해서는 input() 함수를 사용
- 형식 : 변수 = input("문자열")
- input() 함수를 통해 입력받은 값은 모두 문자열로 인식됨

```python
name = input('이름을 입력해주세요')
age = input('나이를 입력해주세요')
print(f"이름은 {name}이고, 나이는 {age} 입니다")
```

```python
# 구구단을 입력받아 실행
dan = input('단을 입력해주세요~')
for number in range(1, 10):
    print(f"{dan} ❎ {number} = {int(dan)*number}")
```

```python
# 구구단을 입력받아 실행 - 시작, 종료 단을 입력받기
s_dan = input('시작할 단수를 입력해주세요~')
e_dan = input('끝낼 단수를 입력해주세요~')
for number in range(1, 10):
    for dan in range(int(s_dan), int(e_dan)+1):
        print(f"{dan} ❎ {number} = {dan*number}", end='\t')
    print()
```
