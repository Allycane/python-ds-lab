# Tuple (튜플)

- 컨테이너 타입은 여러 개의 값을 다루는 자료형입니다. list, tuple, set, dictionary가 있습니다.
    - 리스트(list) : 데이터를 연속적으로 관리, 값을 바꿀 수 있음, 예:['가', '나', '다', '라']

    - 튜플(tuple) : 데이터를 연속적으로 관리, 값을 바꿀 수 없음, 예:('가', '나', '다', '라')

     -세트(set) : 집합과 같은 속성, 예:{'가', '나', '다', '라'}

    - 딕셔너리(dictionary) : key-value 쌍, 예: {'가':123, '나':456, '다':789}

- 리스트와 유사
- 한 번 입력한 후 항목을 변경할 수 없는 특징이다
- 소괄호를 사용해서 생성 또는 괄호 없이 생성
- 튜플의 항목은 숫자 문자열, 불린 리스트, 튜플세트, 세트, 딕셔너리 등을 만들 수 있음

## 3. 튜플 아이템 추가/삭제 - 실행불가❌

```python
fruits_tuple
```

```python
# 새로운 아이템 추가 불가
fruits_tuple.append('lemon')
```

## 4. 튜플을 리스트로 형변환

```python
type(fruits_tuple)
```

```python
fruits_list = list(fruits_tuple)
fruits_list
```

```python
type(fruits_list)
```

```python
# fruits_list 마지막에 'orange' 추가
fruits_list.append('orange')
```

```python
fruits_list
```

```python
fruits_tuple
```

```python
# fruits_tuple 객체의 orange 아이템을 Mango로 변경
basket_tuple = ('apple', 'tomate', 'orange')
basket_list = list(basket_tuple)
basket_list.index('orange')
basket_list[basket_list.index('orange')] = 'mango'
basket_tuple = tuple(basket_list)
basket_tuple
```
