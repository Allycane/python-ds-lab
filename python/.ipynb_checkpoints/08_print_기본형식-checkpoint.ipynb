# print()

## 1. 기본 문법

- print(value, ..., sep='구분기호', end='\n', flush=False)

```python
a = 5
a
```

```python
b = 3
b
```

```python
a+b
```

```python
print(a, b, a+b)
```

```python
print(a)
print(b)
print(a+b)
```

```python
print(1, 2, 3, sep='/') # 1-2-3
```

```python
print(1, 2, 3, sep='/', flush=True) # flush=True 값을 지정하면 출력 후 출력한 데이터를 메모리에서 제거함
```

```python
print(a, end='\t')
print(b, end='\t')
print(a+b)
```

```python
name = '유재석'
age = 54
```

```python
# 1. 콤마로 구분
print('이름 : ', name, '나이 : ', age) # int() - 숫자로 변환 / str() - 문자열로 변환
```

```python
# 2. 접합연산자(+) 사용하여 문자열 연결, 문자열만 가능
print('이름 : '+ name, '나이 : '+ str(age)) # int() - 숫자로 변환 / str() - 문자열로 변환
```

```python
# 3. %로 포맷(구버전)
print('이름 : %s, 나이 : %d' % (name, age))
```

```python
# 4. format() 포맷(신버전)
print('이름 : {}, 나이 : {}' .format(name, age))
```

```python
# 5. f-string (recommanded)
print(f'이름 : {name}, 나이 : {age}')
```

```python
# 여러 값을 출력 "," 콤마로 구분
print(1, 2, 3)
print(1, 2, 3, sep=',')
```

```python
# 줄바꿈 없이 옆으로 출력
lst = [1, 2, 3, 4, 5]
for i in lst :  # 0, 1, 2, 3, 4
    print(i, end=' ')
```

```python
# '*' 문자열을 50번 출력
print('*' * 50)
```
