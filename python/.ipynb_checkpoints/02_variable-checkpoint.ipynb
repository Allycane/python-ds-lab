# 2. 변수 이름을 붙이는 규칙(1)

### 🎯 반드시 지켜야 하는 규칙  

- 숫자로 시작하지 않는다.
- 공백이 포함되지 않는다.
- 기호가 포함되지 않는다.
- 예약어(키워드)는 사용하지 않는다.  

```python
# 잘못된 변수 정의
2var = "파이썬" # 숫자로 시작하여 에러 발생 ❌
```

```python
happy python = "파이썬" # 공백포함으로 에러 발생 ❌
```

```python
happy! = "해피" # 특수기호 포함으로 에러 발생 ❌
```

```python
def = '함수' # 예약어 사용으로 에러 발생 ❌
```

```python
# 사용이 가능한 변수명 - snake_case
happy_python = "파이썬"
print(happy_python)
```

```python
# 사용이 가능한 변수명 - camelCase
happyPython = "파이썬"
print(happyPython)
```

# 3. 변수 이름을 붙이는 규칙(2)

### 🎯 에러 발생은 하지 않지만, 참고해야할 사항  

- 대소문자를 구분한다.
- 한 번에 여러 개의 변수 선언이 가능하다.
- 하나의 값을 여러 변수에 저장할 수 있다.

```python
# 1. 대소문자를 구분
abc = 100
abc
```

```python
ABC = 200
ABC
```

```python
print("abc :", abc, " |", " ABC :", ABC)
```

```python
# 2. 한 번에 여러 개의 변수 선언이 가능
x, y, z = 100, 200, 300
print(x, y, z)
```

```python
name1, name2 = "유재석", "노홍철"
print(name1, name2)
```

```python
# 하나의 값을 여러 변수에 저장 가능
xx = yy = zz = "Apple"
print(xx, yy, zz)
```
