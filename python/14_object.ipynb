# OPC (Orject Oriented Programming : 객체 지향 프로그래밍

- 객체를 만들고 이용할 수 있는 기능을 제공하는 프로그래밍
- Object (객체)
    - 속성과 행위로 구성
    - 속성 : 객체가 가지는 값, 상태 => 변수로 정의
    - 행위 : 객체가 동작하는 기능. 행위 => 함수로 정의
- object(객체) 생성 방법
    - 클래스 선언, 속성(변수), 행위(함수) 정의
    - 클래스는 기본 틀 (설계도)와 같이 툴을 만드는 작업
    - 객체는 클래스를 이용하여 생성되며 인스턴스(instance)라고도 함

### 1. 클래스 선언

- 기본 구조

```python
class 클래스명():
    [변수1] # 클래스 변수
    [변수2]
    # 생성자 함수
    def _init_(self, 인자1, 인자2...):
        <코드 블록>
        ...
    def 함수1(self[, 인자1, 인자2, ..., 인자n]): 클래스 함수
        <코드 블록>
        ...
    def 함수2(self[, 인자1, 인자2, ..., 인자n]):
        <코드 블록>
        ...

````python
# Animal 클래스 생성
class Animal():
    className = 'Animal' # 클래스 변수

    def __init__(self, name, emoji): # 생성자 함수, 맨처음 호출
        self.name = name
        self.emoji = emoji

    def display(self):
        print(f"name : {self.name} | emoji : {self.emoji}")
````

````python
# Animal 클래스의 인스턴스 생성
dog = Animal('해피', '🐶')
````

````python
dog.display()
````

````python
cat = Animal('하양이', '😺')
````

````python
cat.display()
````

````python
class Bicycle():

    def __init__(self, wheelSize, color, name):
        self.wheelSize = wheelSize
        self.color = color
        self.name = name

    def info(self):
        print(f"wheelSize : {self.wheelSize}, color : {self.color}")

    def move(self, speed):
        print(f"{self.name} 속도는 {speed}m/h 입니다")
        
````

````python
myBicycle = Bicycle(50, 'Red', '따릉이')
myBicycle.info()
myBicycle.move(50)
````

### 2. 클래스 상속

````python
# Animal 클래스의 자식 Dog 클래스 정의
class Dog(Animal):
    className = 'Dog'

    def __init__(self, name, emoji, color):
        super().__init__(name, emoji)
        self.color = color

    def display(self):
        print(f"name : {self.name} | emoji : {self.emoji} | color : {self.color}")

    def setName(self, name):
        self.name = name

    def getName(self):
        return self.name
````

````python
happy = Dog('해피', '🐶', '갈색')
happy.display()
print(Dog.className)
happy.setName('초코')
happy.display()
print(f"우리 강아지 이름은 {happy.getName()} 입니다")
````

````python
class Cat(Animal):
    className = "cat"

    def __init__(self, name, emoji, color, age, species):
        super().__init__(name, emoji)
        self.color = color
        self.age = age
        self.species = species

    def display(self):
        print(f"name : {self.name} | emoji : {self.emoji} | color : {self.color} | age : {self.age} | species : {self.species}")

    def setName(self, name): self.name = name
    def setEmoji(self, emoji): self.emoji = emoji
    def setColor(self, color): self.color = color
    def setAge(self, age): self.age = age
    def setSpecies(self, species): self.species = species

    def getName(self): return self.name
    def getEmoji(self): return self.emoji
    def getColor(self): return self.color
    def getAge(self): return self.age
    def getSpecies(self): return self.species
````

````python
navi = Cat('나비', '😺', 'Yellow', 3, '브리티시 숏헤어')
navi.display()
navi.setName('하양')
print(f"우리 고양이 이름은 {navi.getName()}이고, {navi.getEmoji()}처럼 생겼어. 나이는 {navi.getAge()}살이고, 종은 {navi.getSpecies()}야!")
````
