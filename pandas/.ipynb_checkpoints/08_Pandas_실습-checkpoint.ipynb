### 4. 선택

```python
# '사용일자' 컬럼을 인덱스로 지정하여 df2 객체를 생성
df2 = subway_df.set_index(['사용일자'])
df2
```

```python
print(subway_df.head(3))
print('-'*100)
print(df2.head(3))
```

```python
# index 검색 : loc - Lable 기준, iloc - id가 숫자일 경우
df2.iloc[1]
```

```python
# 20260610 일자 데이터 검색
df2.loc[20260510]
```

```python
# 이반 데이터 검색 : 펜시, 불린, 쿼리
# '노선명' 컬럼의 2호선 데이터 검색
# 불린 인덱싱 조건 : df2['노선명'] == '2호선'
print(df2[df2['노선명'] == '2호선']['노선명'].count())
```

```python
# 이반 데이터 검색 : 펜시, 불린, 쿼리
# '노선명' 컬럼의 2호선 데이터 검색
print(df2.query('노선명 == "2호선"')['노선명'].count())
```

```python
# '승차총승객수'가 50000명 이상인 '역명' 조회
df2[df2['승차총승객수'] >= 50000]['역명'].unique()
```

```python
df2.query('승차총승객수 >= 50000')['역명'].unique()
```

```python
# '승차총승객수'가 50000명 이상이고, '역명' 이름이 '강남'인 데이터 조회
print(df2.query('승차총승객수 >= 50000 & 역명 == "강남"').count())
```

```python
# '승차총승객수'가 50000명 이상이거나, '역명' 이름이 '강남'인 데이터 조회
print(df2.query('승차총승객수 >= 50000 | 역명 == "강남"').count())
```

```python
# '승차총승객수' 컬럼의 최댓값
print(df2['승차총승객수'].max(), '명')
```

```python
# '승차총승객수' 컬럼의 최솟값
print(df2['승차총승객수'].min(), '명')
```

```python
# '승차총승객수' 컬럼의 최댓값, '하차총승객수' 컬럼의 최솟값을 조회
df2.agg(
    {
        '승차총승객수' : max,
        '하차총승객수' : min
    }
)
```

```python
# '승차총승객수', '하차총승객수' 컬럼의 최댓값, 최솟값, 평균값을 조회
df2[['승차총승객수', '하차총승객수']].agg(['max', 'min', 'mean', 'median']).round(0)
```

### 6. 연산

```python
# '승차총승객수' 컬럼에 10을 더하여, '승차총승객수_10' 컬럼 생성
subway_df['승차총승객수_10'] = subway_df['승차총승객수'] * 10
```

```python
subway_df
```

```python
# '승차총승객수' - '하차총승객수' 결과를 '승하차총승객수' 컬럼으로 생성하여 추가
subway_df['승하차총승객수'] = subway_df['승차총승객수'] - subway_df['하차총승객수']
subway_df
```

```python
# '승차총승객수' + '하차총승객수' 결과를 '승하차총승객수합계' 컬럼으로 생성하여 추가
subway_df['승하차총승객수합계'] = subway_df['승차총승객수'] + subway_df['하차총승객수']
subway_df
```

```python
# '승하차총승객수' 의 평균값, 중앙값
subway_df['승하차총승객수'].agg(['mean', 'median']).round()
```

```python
# index 재생성 후 df2에 반영, 기존 index 컬럼 유지
df2.reset_index(drop=False, inplace=True)
```

```python
df2.head()
```

```python
# '노선명' 컬럼의 2호선 데이터를 추출하여 df2_rail2 객체 생성
df2_rail2 = df2.query('노선명 == "2호선"')
df2_rail2.head()
```

```python
# '노선명' 컬럼의 2호선이고, '역명'이 '강남'인 데이터를 추출하여 df_rail2_gn 객체 생성
df_rail2_gn = subway_df.query("노선명 == '2호선' & 역명 == '강남'")
df_rail2_gn.head()
```

```python
# df2_rail2 데이터프레임과 df_rail2_gn 데이터프레임을 사칙연산 - index와 column 명이 모두 동일해야 연산이 가능함
print(df2_rail2 + df_rail2_gn)
```

```python
# df2_rail2 데이터프레임과 df_rail2_gn 데이터프레임을 사칙연산 - inner join
# pandas.merge
pd.merge(df2_rail2, df_rail2_gn)
```

```python
df_rail2_gn.drop(['승차총승객수_10', '승하차총승객수합계'], axis=1, inplace=True)
df_rail2_gn
```

```python
pd.merge(df2_rail2, df_rail2_gn)
```

```python
# left outer join
pd.merge(df2_rail2, df_rail2_gn, how='left')['역명'].head()
```

```python
# right outer join
pd.merge(df2_rail2, df_rail2_gn, how='right')['역명'].head()
```

```python
td = pd.read_csv('data/titanic/train.csv')
td
```

```python
td.info()
```

```python
td.isna().sum()
```

```python
# Age 컬럼은 결측치를 평균값으로 대체
age_mean = int(td['Age'].mean())
td['Age'] = td['Age'].fillna(age_mean)
```

```python
td.isna().sum()
```

```python
# Cabin 컬럼 삭제
td.drop('Cabin', axis=1, inplace=True)
```

```python
td.info()
```

```python
# Embarked 결측치는 이전값으로 대체
td['Embarked'] = td['Embarked'].ffill()
```

```python
td.isna().sum()
```

### 12. 시각화(matplotlib)

```python
import matplotlib.pyplot as plt
import matplotlib
```

```python
# 노선별 승차총승객수가 많은 노선 리스트 조회
line_sum = subway_df.groupby('노선명')['승차총승객수'].sum().sort_values(ascending=False)
```

```python
line_sum
```

```python
# 막대 그래프
plt.figure(figsize=(14, 5))
plt.bar(line_sum.index, line_sum.values, color='steelblue')
plt.title('노선별 승차 총 승객 수 합계')
plt.xlabel('노선명')
plt.ylabel('승차 총 승객수')
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
# '승차총승객수' 기준, 상위 10개 역 막대 그래프로 시각화
top10_station = subway_df.sort_values('승차총승객수', ascending=False)[['역명', '승차총승객수']].head(10)
top10_station
```

```python
# 막대 그래프
plt.figure(figsize=(14, 5))
plt.bar(top10_station['역명'], top10_station['승차총승객수'], color='midnightblue')
plt.title('Top 10 Station')
plt.xlabel('Station Name')
plt.ylabel('Passangers')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
