# Pandas 실습2

### 1. 데이터 준비

```python
import warnings
warnings.filterwarnings(action='ignore')

import pandas as pd
import numpy as np
```

```python
df1 = pd.read_csv('data/서울특별시 공공자전거 일별 대여건수_25.7-12.csv')
df1
```

```python
df2 = pd.read_excel('data/대여소별 대여내역(2017년).xlsx')
df2
```

```python
# df1의 일자별 대여건수 top10
df1.shape
df1['대여일자'].dtypes
```

```python
# '대여일자'의 데이터타입 변환, astype()
df1['대여일자'] = pd.to_datetime(df1['대여일자'])
df1.dtypes
df1['Year'] = df1['대여일자'].dt.year
df1['Month'] = df1['대여일자'].dt.month
df1['Day'] = df1['대여일자'].dt.day
df1
```

```python
df1['Month'].value_counts()
```

```python
# df1의 2025년도 월별 대여건수
df1_rentals = df1.groupby('Month')['Month'].sum()
df1_rentals.values
```

```python
import matplotlib.pyplot as plt
import matplotlib
```

```python
plt.figure(figsize=(14, 5))
plt.bar(df1_rentals.index, df1_rentals.values, color='steelblue')
plt.title('Bicycle Rental of 2025"s')
plt.xlabel('Month')
plt.ylabel('Rental')
plt.xticks(rotation=45, ha='right')
plt.tight_layout
plt.show()
```

- 요일별 대여건수 그래프 시각화

```python
df1
```

```python
df1['대여일자'].dt.day_name()
```

```python
# 0: 월요일, 1: 화요일, 2: 수요일...
# 'weekday' 컬럼 생성하고 요일을 저장
df1['대여일자'].dt.weekday
```

```python
df1['weekday'] = df1['대여일자'].dt.weekday
```

```python
df1
```

```python
# 요일을 매핑하는 딕셔너리 객체 생성
week_dict = {
    0: '월',
    1: '화',
    2: '수',
    3: '목',
    4: '금',
    5: '토',
    6: '일'
}
df1['weekday'] = df1['대여일자'].dt.weekday.map(week_dict)
df1
```

```python
# 요일별 대여건수 그래프 작성
order = ['월', '화', '수', '목', '금', '토', '일']
df1['weekday'].value_counts()
result = df1.groupby('weekday')['대여건수'].sum().reindex(order)
result
```

```python
plt.figure(figsize=(10, 5))
plt.bar(result.index, result.values, color="red")
plt.title('Rental Rating of WeekDay')
plt.xlabel('Day')
plt.ylabel('Rental Count')
plt.show()
```

##### 월별 전체 대여건수 Top10

```python
df2.shape
df2.columns = ['그룹', '대여소', '일자', '건수']
```

```python
df2.dtypes
```

```python
df2['Year'] = df2['일자'].dt.year
df2['Month'] = df2['일자'].dt.month
df2['Day'] = df2['일자'].dt.day
```

```python
df2_rental = df2.groupby('Month')['건수'].sum()
df2_rental
```

```python
# plt.figure(figsize=(20, 5))
# plt.bar(df2_rental.index, df2_rental.values, color='brown')
df2_rental.plot(
    kind='line',
    marker='o',
    figsize=(20, 5)
)
plt.title('Bicycle Rental of 2017"s')
plt.xlabel('Month')
plt.ylabel('Rental')
plt.grid()
plt.xticks(rotation=45, ha='right')
plt.tight_layout
plt.show()
```

```python
df2.head()
```

```python
df2['대여소'].value_counts()
station_top5 = df2.groupby('대여소')['건수'].sum().sort_values(ascending=False).head(5)
station_top5
```

```python
plt.figure(figsize=(10, 5))
plt.bar(station_top5.index, station_top5.values, color='beige')
plt.title('Top5 Rental Rating of Center')
plt.xlabel('Station_Name')
plt.ylabel('Rental Counts')
plt.xticks(rotation=45, ha='right')
plt.tight_layout
plt.show()
```

##### 그룹별 대여건수 Top10

```python
df2.dtypes
```

```python
group_top10 = df2.groupby('그룹')['건수'].sum().sort_values(ascending=False).head(10)
group_top10
```

```python
plt.figure(figsize=(10, 5))
plt.bar(group_top10.index, group_top10.values, color='yellow')
plt.title('Top10 Rental Rating of City')
plt.xlabel('City Group')
plt.ylabel('Rental Counts')
plt.xticks(rotation=45, ha='right')
plt.tight_layout
plt.show()
```

##### 대여소별 평균 대여건수 Top15

```python
rental_mean = df2.groupby('대여소')['건수'].mean().sort_values(ascending=False).head(15).round(2)
rental_mean
```

```python
plt.figure(figsize=(20, 7))
plt.bar(rental_mean.index, rental_mean.values, color='darkgray')
rental_mean.plot(
    kind='barh',
    figsize=(20, 7),
    color='red'
)
plt.title('Top15 Mean of Rental Counts of Station')
plt.xlabel('Station Name')
plt.ylabel('Mean of Rental Counts')
plt.grid()
# plt.xticks(rotation=45, ha='right')
# plt.tight_layout
plt.show()
```
