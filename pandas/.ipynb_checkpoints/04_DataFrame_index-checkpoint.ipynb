# DataFrame Index

## Index란?
- Pandas의 Index는 DataFrame과 Series의 각 행(row)를 고유하게 식별하는 역할을 수행함
- 관계형 Database의 Primary key(pk)와 유사한 개념

- 특징
    - 데이터 식별 역할 수행
    - 숫자, 문ㅈ열도 가능
    - 생성 후 직접 수정 불가
    - reset_index() 함수를 이용해야 재생성 가능

## 2. Index 기반 연산 방식
- ✨판다스는 인덱스 이름을 기준으로 데이터를 맞춘 후 연산함

### (1) Series 연산
- ✨ 동일한 인덱스를 가진 데이터끼리만 연산됨

#### 기본 Series 생성 및 연산

```python
# Series 생성
s1 = pd.Series([10, 20, 30])
s2 = pd.Series([40, 50, 60])
print(s1, '\n', s2)
```

```python
# Series 사칙 연산
print(s1 + s2)
print(s1 - s2)
print(s1 * s2)
print(s1 / s2)
```

#### index 지정

```python
s1 = pd.Series([10, 20, 30], index=['a', 'b', 'c'])
s2 = pd.Series([40, 50, 60], index=['b', 'c', 'd'])
print(s1, '\n', s2)
```

```python
# Series 사칙연산 - 동일한 인덱스 기준으로 연산이 진행됨
print(s1 + s2)
print(s1 - s2)
print(s1 * s2)
print(s1 / s2)
```

### (2) DataFrame 연산
- ✨동일한 인덱스, 동일한 컬럼명 기준으로 연산이 진행됨

#### DataFrame 생성

```python
df1 = pd.DataFrame(
    {
        'col1' : [10, 20, 30],
        'col2' : [20, 30, 40]
    },
    index=['a', 'b', 'c']
)
```

```python
df2 = pd.DataFrame(
    {
        'col1' : [10, 20, 30],
        'col2' : [20, 30, 40]
    },
    index=['a', 'b', 'c']
)
```

```python
print(df1)
print()
print(df2)
```

```python
# 인덱스, 컬럼명이 일치해야만 연산이 수행됨
df1 + df2
```

```python
# df1 객체에 col3 컬럼 추가, 값은 100
df1['col3'] = 100
```
