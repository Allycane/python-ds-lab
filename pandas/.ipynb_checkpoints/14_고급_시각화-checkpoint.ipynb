### seaborn = barplot : 범주형 데이터

```python
df = sns.load_dataset('tips')
print('tips 데이터 앞 3행:')
print(df.head(3))

sns.set_theme(style='whitegrid', rc={'figure.figsize': (10, 4)}, font='Malgun Gothic')

fig, axes = plt.subplots(1, 2)

# 요일별 평균 청구금액
sns.barplot(data=df, x='day', y='total_bill', estimator='mean',
            hue='day', legend=False, ax=axes[0])
axes[0].set_title('요일별 평균 청구금액')

# 요일별 성별 평균 청구금액
sns.barplot(data=df, x='day', y='total_bill', estimator='mean',
            hue='sex', ax=axes[1])
axes[1].set_title('요일별 성별 평균 청구금액')

plt.tight_layout()
plt.show()
```

## seaborn lineplot - 월별 항공 탑승객

```python
flights.head()
```

```python
# [예제 2] seaborn lineplot + 누적 막대그래프 (code8_2)
import seaborn as sns
import matplotlib.pyplot as plt

flights = sns.load_dataset('flights')
sns.set_theme(style='whitegrid', rc={'figure.figsize': (10, 4)}, font='Malgun Gothic')
sns.set_palette('hls', 12)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# 선그래프
sns.lineplot(data=flights, x='year', y='passengers',
             hue='month', ax=axes[0])
axes[0].set_title('연도별 월별 항공 탑승객 수')
axes[0].legend(loc='upper left', fontsize=7)

# 누적 막대그래프
df2 = flights.pivot_table(index='year', columns='month',
                          values='passengers', aggfunc='mean')
df2.plot.bar(stacked=True, ax=axes[1], legend=False)
axes[1].set_title('연도별 누적 탑승객 (월별 색상)')

plt.tight_layout()
plt.show()
```

### [예제 3] seaborn boxplot — 서울 기온

```python
# 서울 월별 기온 분포 - seaborn boxplot
df = pd.read_csv('data/OBS_ASOS_DD_20250101.csv')
df.head()
```

```python
df.info()
```

```python
df['일시'] = pd.to_datetime(df['일시'])
```

```python
# '일시' 컬럼을 year, month, day 컬럼으로 분리
df['연도'] = df['일시'].dt.year
df['월'] = df['일시'].dt.month
df['일'] = df['일시'].dt.day
df.head()
df['월'].value_counts()
```

```python
sns.set_theme(style='white', rc={'figure.figsize': (10, 5)}, font='Malgun Gothic')
sns.boxplot(
            data=df, 
            x='월', 
            y='평균기온(°C)',
            legend=False
)
plt.title('서울 월별 기온 분포 (2023년)')
plt.xlabel('월')
plt.ylabel('평균기온 (°C)')
plt.show()
```

```python
# [ 실습 ] 2016 ~ 2025년 까지의 평균 기온, 평균 습도를 boxplot으로 생성하여 한 행에 두개 그래프 생성
# 년도 기준 그루핑
# x = 년도, y=평균기온, 평균습
df = pd.read_csv('data/OBS_ASOS_MNH_20260623133953.csv')
df.head()
```

```python
df.info()
```

```python
df['일시'] = pd.to_datetime(df['일시'])
df['일시'].head()
```

```python
df['연도'] = df['일시'].dt.year
df['월'] = df['일시'].dt.month
df['일'] = df['일시'].dt.day
df.head()
df['연도'].value_counts()
```

```python
sns.set_theme(style='white', rc={'figure.figsize': (12, 5)}, font='Malgun Gothic')
fig, axes = plt.subplots(1, 2)

sns.boxplot(
            data=df, 
            x='연도', 
            y='평균기온(°C)',
            hue='연도',
            legend=False,
            ax=axes[0]
)
axes[0].set_title('평균기온')

sns.boxplot(
            data=df, 
            x='연도', 
            y='평균상대습도(%)',
            hue='연도',
            legend=False,
            ax=axes[1]
)
axes[1].set_title('평균습도')
plt.show()
```

### [예제 4] seaborn scatterplot 버블 차트 — 범죄율

```python
# [예제 4] 버블 차트 — 미국 주별 범죄율 (code8_4)
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('data/crimeRatesByState2005.csv')
print('컬럼:', df.columns.tolist())
print(df.head(3))

sns.set_theme(rc={'figure.figsize': (12, 10)})
sns.scatterplot(data=df, x='murder', y='burglary',
                size='population', sizes=(20, 3000),
                hue='state', alpha=0.5, legend=False)
plt.xlim(0, 12)
plt.title('미국 주별 살인율 vs 절도율 (원 크기: 인구수)')

# 주 이름 레이블
for i in range(len(df)):
    plt.text(df.murder.iloc[i], df.burglary.iloc[i], df.state.iloc[i],
             ha='center', size='x-small', color='dimgray')
plt.show()
```

### [예제 5] mosaic plot — 타이타닉 생존율

```python
# [예제 5] Mosaic Plot — 타이타닉 남녀 생존율 (code8_5)
from statsmodels.graphics.mosaicplot import mosaic
import platform

if platform.system() == 'Windows':
    plt.rcParams['font.family'] = 'Malgun Gothic'
elif platform.system() == 'Darwin':
    plt.rcParams['font.family'] = 'AppleGothic'
plt.rcParams['axes.unicode_minus'] = False

df = sns.load_dataset('titanic')
df = df.replace({'survived': {0: '사망', 1: '생존'}})
df = df.replace({'sex': {'male': '남성', 'female': '여성'}})

props = lambda key: {'color': 'lightgreen' if '생존' in key else 'gold'}

mosaic(data=df.sort_values('sex'),
       index=['sex', 'survived'],
       properties=props,
       axes_label=True,
       title='타이타닉 남녀 생존 비율')
plt.show()
```

### [예제 6] seaborn heatmap — 항공 탑승객

```python
# [예제 6] 히트맵 — 연도×월별 항공 탑승객 (code8_6)
flights = sns.load_dataset('flights')

pivot = flights.pivot_table(index='month', columns='year',
                            values='passengers', aggfunc='mean')
pivot
```

```python
sns.set_theme(rc={'figure.figsize': (10, 6)}, font='Malgun Gothic')
sns.heatmap(pivot, annot=True, fmt='.0f', cmap='YlOrRd', linewidths=0.5)
plt.title('항공 탑승객 수 히트맵 (연도×월)', fontsize=14)
plt.tight_layout()
plt.show()
```

### [예제 7] plotly treemap — 국가별 인구/GNI

```python
from statsmodels.graphics.mosaicplot import mosaic
import platform
```

```python
df = sns.load_dataset('titanic')
df = df.replace({'survived': {0: '사망', 1: '생존'}})
df = df.replace({'pclass': {1: '1등실', 2: '2등실', 3: '3등실'}})

df.head()
```

```python
# 모자이크 플롯
props = lambda key: {'color': 'lightgreen' if '생존' in key else 'gold'}
mosaic(data=df.sort_values('pclass'), index=['pclass', 'survived'],
       properties=props, axes_label=True,
       title='타이타닉 선실별 생존 비율')
plt.show()
```

### 실습 2. heatmap — iris 상관계수 행렬 시각화

**조건**

1. `iris.csv`의 수치형 컬럼 4개의 **상관계수 행렬**을 계산한다
2. `sns.heatmap()`으로 시각화한다
3. 각 셀에 상관계수 값을 표시하고(`annot=True`), 색상 팔레트는 `'coolwarm'`을 사용한다
4. 대각선 색상을 통해 완전 상관(`1.0`)임을 확인하고 Markdown 셀에 해석 한 줄을 작성한다

```python
df = pd.read_csv('data/iris.csv')
df.head(3)

# 수치형 데이터만 추출
df_num = df.drop(columns='Species')
df_num.head()
```

```python
# 상관계수 행렬 계산
df_corr = df_num.corr().round(2)
df_corr
```

```python
# 상관계수 행렬에 대한 히트맵 생성
sns.set_theme(rc={'figure.figsize': (10, 6)}, font='Malgun Gothic')\

sns.heatmap(df_corr, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('IRIS 상관계수 행렬에 대한 히트맵', fontsize=14)
plt.tight_layout()
plt.show()
```

### 실습 3. 타이타닉 성별×객실등급 생존 분석 (seaborn + mosaic)

**조건** (seaborn 내장 titanic 사용)

1. 성별(`sex`)×객실등급(`pclass`)별 생존율(survived 평균)을 `barplot`으로 그린다
2. `sns.catplot()`을 이용해 col='sex'로 분할하여 객실등급별 생존율을 비교한다
3. 아래 해석을 Markdown 셀에 작성한다: 어떤 그룹의 생존율이 가장 높은가?

```python
df = pd.read_csv('data/titanic/train.csv')
df.head()
```

```python
df_survived = df.groupby(['Sex', 'Pclass'])['Survived'].mean().round(2).reset_index()
df_survived
```

```python
sns.set_theme(style='whitegrid', rc={'figure.figsize': (7, 5)}, font='Malgun Gothic')
# set_theme 실행 시 자체 font 타입으로 리셋되므로 font 형식을 추가해야 함.

sns.barplot(
            data=df_survived, 
            x='Sex', 
            y='Pclass', 
            estimator='mean',
            hue='Sex', 
            palette='Set2',
            legend=False
)

plt.title('성별×객실등급 생존 분석')
plt.xlabel('성별')
plt.ylabel('객실번호')

plt.show()
```
