```python
# 1반의 수학, 과학 점수 평균조회
exam_df.query('nclass == 1')[['math', 'science']].mean()
```

## 2. Merge
- SQL의 JOIN과 유사함
- inner join
- left join
- right join
- outer join

### (1) inner join
- key를 기준으로 두 개의 DataFrame(Table)에 동일하게 존재하는 데이터를 추출
- 예제) '이름'을 키로 하여 inner join(교집합)

```python
df1 = pd.DataFrame(
    {
        '이름' : ['유재석', '정형돈', '노홍철'],
        '키' : [176, 175, 174]
    }
)
```

```python
df2 = pd.DataFrame(
    {
        '이름' : ['유재석', '정형돈', '장윤주'],
        '성별' : ['남자', '남자', '여자']
    }
)
df2
```

```python
# inner join how = 'inner' 기본값
pd.merge(df1, df2)
```

### (2) left outer join
- 왼쪽 에디터셋에 있는 데이터프레임 (테이블)의 키를 기준으로 병합

```python
#df1, df2 데이터프레임 left outer join(inner join + left) 테이블 모든 데이터셋
pd.merge(df1, df2, how= 'left')
```

### (3) Right Outer Join
- 오른쪽 에디터셋에 있는 데이터프레임 (테이블)의 키를 기준으로 병합

```python
# df1, df2 데이터프레임 Right Outer Join (inner join + right) 테이블 모든 데이터셋
pd.merge(df1, df2, how='right')
```

### (4) Full Outer Join
- 키를 기준으로 두 데이터프레임의 존재하는 모든 데이터 기준으로 병함
  

```python
# full outer join
pd.merge(df1, df2, how='outer')
```

## 3. Concat() 함수를 이용한 데이터 병합

```python
df1 = pd.DataFrame(
    {
        '이름' : ['유재석', '정형돈', '노홍철'],
        '키' : [176, 175, 174],
        '나이' : [54, 32, 27],
        '주소' : ['압구정동', '개화동', '홍대']
    }
)
df1
```

```python
df2 = pd.DataFrame(
    {
        '이름' : ['강호동', '이수근', '김C'],
        '키' : [180, 164, 172],
        '나이' : [59, 45, 60],
        '주소' : ['미국', '캐나다', '호주']
    }
)
df2
```

```python
# df1, df2를 행 기준으로 concat() - 새로운 객체로 생성됨
df_concat = pd.concat([df1, df2], axis=0)
df_concat
```

```python
# df_concat 데이터프레임 인덱스 재생성, 기존 인덱스 삭제, 원본 반영
df_concat.reset_index(drop=True, inplace=True)
df_concat
```

```python
# df1, df2를 열 기준으로 concat
df_concat1 = pd.concat([df1, df2], axis=1)
df_concat1
```

```python
# df1 객체에 '직업' 컬럼 추가
df1['직업'] = '개발자'
df1
```

```python
pd.concat([df1, df2])
```

```python
join_concat = pd.concat([df1, df2], join='inner')
join_concat.reset_index(drop=True, inplace=True)
join_concat
```
