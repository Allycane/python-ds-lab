# DataFrame과 List, Dictionary, ndarray 상호 변환

## 1. 1차원 데이터를 Dataframe으로 변환

```python
import warnings
warnings.filterwarnings(action='ignore')

import pandas as pd
import numpy as np
```

```python
# 1차원 리스트 생성
col_name1 = ['col_1']
list1 = [1, 2, 3]
print(list1, type(list1))
```

```python
# ndarray로 변환
array1 = np.array(list1)
print(array1, type(array1))
```

```python
# list1, array1을 dataframe으로 변환(생성)
df1_1 = pd.DataFrame(list1, columns = col_name1, index = (0, 0 ,0))
df1_2 = pd.DataFrame(array1, columns = col_name1)
print(df1_1)
print('-'*30)
print(df1_2)
```

## 2. 2차원 데이터를 DataFrame으로 변환

```python
# 컬럼명 생성
col_names = ['col1', 'col2', 'col3']
col_names
```

```python
# 2차원 리스트 생성
list2 = [[1, 2, 3],[4, 5, 6]]
```

```python
# list2 ndarray로 변환
array2 = np.array(list2)
array2
```

```python
# list2, arrya2를 dataframe으로 변환
df2_1 = pd.DataFrame(list2, columns = col_names)
df2_2 = pd.DataFrame(array2, columns = col_names)
print(df2_1)
print('-'* 30)
print(df2_2)
```

## 3. 딕셔너리를 DataFrame으로 변환

```python
# dictionary 생성
dict1 = {
    'col1' : [1, 2, 3],
    'col2' : [4, 5, 6],
    'col3' : [7, 8, 9]
}
dict1
```

```python
# dataframe으로 dictionary 변환
df_dict1 = pd.DataFrame(dict1)
df_dict1
```

## 4. DataFrame을 ndarray, list, dictionary로 변환

```python
# 각 df_dict1f를 array, list로 전합 / DataFrame.value > Numpy.tolist() > list 
list3 = df_dict1.values.tolist()
print(type(list3), '\n', list3)
```

```python
# df_dict1 데이터프레임을 dictionary로 변환
dict3 = df_dict1.to_dict()
print(type(dict3), '', dict3)
```

## 5. shape 속성 정리

- python List : Attribute Error
- python Dictionary : Attribute Error
- Numpy ndarray : 사용 가능
- Pandas DataFrame : 사용 가능
- Pnadas Series : 사용 가능

```python
# List 생성
print(type(list1), list1)
print(f"shape : {list1.shape}") # ❌ 'list object'는 shape라는 속성을 가지고 있지 않음
```

```python
# Dictionary 생성
print(type(dict1), dict1)
print(f"shape : {dict1.shape}") # ❌ 'dict object'는 shape라는 속성을 가지고 있지 않음
```

```python
# array
print(type(array1), array1)
print(f"shape : {array1.shape}")
```

```python
# DataFrame
print(type(df_dict1), df_dict1)
print(f"shape : {df_dict1.shape}")
```

## 6. DataFrame 컬럼 데이터 생성 및 수정

```python
# 데이터셋 수치 정보 조회
titanic_df.describe() # .describe()도 외워야 함
# 수치 데이터만 출력하기 때문에 int타입 혹은 float타입의 컬럼만 출력됨
```

```python
# Age_0 컬럼 추가
# titanic_df.info()
titanic_df['Age_0'] = 0 # 컬럼이 존재하면 값을 수정한다, 그렇지 않을 경우 새로운 컬럼을 추가한다.
```

```python
titanic_df.info()
```

```python
# 일부 컬럼 조회
# titanic_df['Age', 'Age_0']
```

```python
# Family_no 컬럼 추가, SibSp 컬럼 + Parch 컬럼 값 합계 + 1
titanic_df.info()
titanic_df['Family_no'] = titanic_df['SibSp'] + titanic_df['Parch'] + 1
titanic_df.info()
```

## 7. DataFrame 데이터 삭제
- drop(컬럼리스트, axis)

```python
titanic_df.info()
```

```python
# Age_0 삭제
td_drop = titanic_df.drop('Age_0', axis=1) # axis=0 행 기준 실행 시 Error 발생, axis=1로 실행해야 함
```

```python
titanic_df.head() # Age_0 객체 존재
```

```python
td_drop.head() # Age_0 객체 존재하지 않음
```

```python
# 원본 데이터셋 (titanic_df)의 Age_0 컬럼 삭제 후 반영
titanic_df.drop('Age_0', axis=1, inplace=True)
```

```python
titanic_df.info() # Age_0 컬럼 삭제 -> 원본에 반영됨
```

```python
# 다중 컬럼 삭제 - Age_by_10 + family_no
titanic_df.drop(['Age_by_10', 'Family_no'], axis=1, inplace=True)
```
