# 결손 데이터 (결측치) 처리

### 결손데이터 (결측치) 란?
- 결측치(Missing Data)는 값이 존재하지 않는 데이터를 의미
- Pandas에서는 NaN으로 표현됨

** 결측치가 있다면??? 어떻게 처리할 것인가?? **
1) 실무자와 협의한 후 진행 - 제거도 한 방법!!
2) 평균으로 대체하는 값, 중앙값으로 대체한느 값..
    예) 날짜마다 온도가 저장되어 있는 ㄷ이터셋이라면..
   20260604    20
   ...
   20260614    32
   20260615    NaN => 앞이나 뒤의 값으로 대체함
   20260616    31
   ...
   20260630    35


```python
import warnings
warnings.filterwarnings(action='ignore')

import pandas as pd
import numpy as np
```

### 1. 결손데이터 생성

```python
df = pd.DataFrame(
    {
        'Name' : ['Smith', 'Lee', 'Kate', 'Catherine'],
        'Age' : [20, np.nan, 30, 40],
        'Score' : [90, 85, np.nan, 75]
    }
)
df
```

```python
df.shape
```

```python
df.info()
```

```python
df.describe()
```

### 2. 결손 데이터 확인
- isna() = isnull()

```python
df.isnull()
```

```python
df.isnull().count()
```

```python
# NaN값이 True인 컬럼만 합계로 계산
print(df.isnull().sum())
print(df.isna().sum())
```

### 3.  결손 데이터 삭제
- dropna() : 결손 데이터가 포함된 행 또는 열 삭제

```python
filter_na = df.dropna()
```

```python
print(df)
print('-' * 30)
print(filter_na)
```

```python
# 결측치 데이터가 포함된 열 삭제
filter_na_cols = df.dropna(axis=1)
```

```python
print(df)
print('-' * 30)
print(filter_na_cols)
```

```python
# 결측치가 저장된 특정 컬럼만 삭제
df.dropna(axis=0, subset=['Age'])
```

```python
# Addr 컬럼 추가, 값은 모두 결측치
df['Addr'] = np.nan
```

```python
df
```

```python
# 모든 데이터가 결측치인 컬럼 삭제
df.dropna(how='all', axis=1, inplace=True)
df
```

```python
# 하나라도 결측치를 포함한 경우, 삭제
df.dropna(how='any', axis=0, inplace=False)
```

## 4. 결손 데이터 대체
- fillna() : 결손 데이터를 특정값으로 대체

```python
df
```

```python
# 결손 데이터를 0으로 대체
df.fillna(0)
```

```python
# 결손 데이터를 unknown으로 대체
df_str = df.fillna('unknown')
df_str.info()
```

```python
# Age 컬럼의 결손 데이터만 0으로 대체
df[['Age', 'Score']].fillna(0)
```

```python
df
```

```python
df['Age'] = df['Age'].fillna(0)
df
```

```python
# Age 컬럼의 결측치를 평균값으로 대체
age_mean = int(df["Age"].mean())
age_mean
```

```python
df['Age'].fillna(age_mean)
df
```

```python
# Score 결ㅊ측치를 평균값으로 대체
df['Score'].fillna(df['Score'].mean().round(2))
```

## 6. 최빈값으로 대체
- 가장 많이 등장하는 값으로 대체
- mode() : 가장 많이 등장한 값 반환

```python
df = pd.DataFrame(
    {
        'Name' : ['Smith', 'Lee', 'Kate', 'Catherine'],
        'Age' : [20, np.nan, 30, 40],
        'Score' : [90, 85, np.nan, 75]
    }
)
```

```python
# Age 컬럼의 최빈값 출력
df['Age'].mode()[2]
```

```python
# Age 컬럼의 결측치를 최빈값으로 대체
df['Age'] = df['Age'].fillna(df['Age'].mode()[0])
df
```

### 7. 이전 데이터로 대체
- ffill : forward fill

```python
df = pd.DataFrame(
    {
        'Name' : ['Smith', 'Lee', 'Kate', 'Catherine'],
        'Age' : [20, np.nan, 30, 40],
        'Score' : [90, 85, np.nan, 75]
    }
)
df
```

```python
# Age, Score 컬럼의 결측치를 ffill 함수로 대체
df.ffill()
```

### 8. 다음 데이터로 대체
- bfill : backward fill

```python
df.bfill()
```

```python
# Age 컬럼을 다음 값으로 대체
df['Age'] = df['Age'].bfill()
df
```

### 9. Titanic 데이터 결측치 처리

```python
titanic_df = pd.read_csv('data/titanic/train.csv')
titanic_df.head()
```

```python
titanic_df.info()
```

```python
# 결손데이터가 몇개인지 확인
titanic_df.isnull().sum()
```

```python
# Cabin 컬럼의 687개의 결측 데이터는 '0000'으로 대체
titanic_df['Cabin'] = titanic_df['Cabin'].fillna('0000')
titanic_df.head(3)
```

```python
titanic_df.isna().sum()
```

```python
# Age 컬럼은 평균값을 구해 겨측 데이터를 대체
titanic_df['Age'] = titanic_df['Age'].fillna(titanic_df['Age'].mean())
titanic_df
```

```python
titanic_df.isna().sum()
```

```python
# Embarked의 결측치를 최빈값으로 대체
titanic_df['Embarked'] = titanic_df['Embarked'].fillna(titanic_df['Embarked'].mode()[0])
titanic_df.isna().sum()
```

### 10. nunique & replace

#### (1) nunique

```python
# value_counts() => 범주형 데이터 조회
titanic_df['Embarked'].value_counts()
```

```python
# titanic_df['Name'].value_counts()
print(titanic_df['Name'].nunique())
print(titanic_df['Pclass'].nunique())
print(titanic_df['Survived'].nunique())
```

#### (2) replace
- 원본값을 다른값으로 대체

```python
# Cabin의 '0000' 값을 'C000' 값으로 대체
titanic_df['Cabin'] = titanic_df['Cabin'].replace('0000', 'C000')
titanic_df.head()
```
