```python
import pandas as pd
```

저장된 모델을 불러서 예측하기
- SepalLength : 5.1
- SepalWidth : 3.0
- PetalLength : 1.3
- PetalWidth : 0.2

```python
data = {
    'SepalLength' : [5.1],
    'SepalWidth' : [3.0],
    'PetalLength' : [1.3],
    'PetalWidth' : [0.2]
}

df = pd.DataFrame(data)
df
```

```python
import joblib
```

```python
# 모델 불러오기
svm_iris = joblib.load('../Data/svm_iris.h5')
rf_iris = joblib.load('../Data/rf_iris.h5')
knn_iris = joblib.load('../Data/knn_iris.h5')
```

```python
# pred = knn.predict([[3.4, 3.7, 0.5]])
# print(f"추천하는 포지션은 {pred[0]} 입니다.")

svm_pred = svm_iris.predict(df)
print(f"SVM 모델 : 해당하는 붓꽃은 {svm_pred[0]} 입니다.")
```

```python
rf_pred = rf_iris.predict(df)
print(f"RandomForest 모델 : 해당하는 붓꽃은 {rf_pred[0]} 입니다.")
```

```python
knn_pred = knn_iris.predict(df)
print(f"kNN 모델 : 해당하는 붓꽃은 {knn_pred[0]} 입니다.")
```
