## 머신러닝이란?
: 수 많은 데이터를 "학습"시켜 안에 있는 패턴( 규칙 )을 찾아내고, 
이를 기반으로 데이터를 분류( Classification )하거나 예측( Predict )함.

- 일반 소프트웨어( 결정론적 소프트웨어 )는 언제나 똑같은 과정을 거쳐 똑같은 결과를 낸다.
- 머신러닝 기반 소프트웨어 ( 비결정론적 소프트웨어 )는 데이터와 알고리즘에 따라 때로는 다른 결과를 추론해낸다.
    ( 추론 : 이미 알고 있는 것으로부터 논리적 결론을 도출하는 과정 )

## 작업 방법
- 어느정도 규묘가 있는 샘플 데이터를 입력으로 넣어 분석 시킨다.
- 이 때 데이터에서 일정한 규칙을 찾아낸다.
- 찾아낸 규칙을 기반으로 다른 데이터를 분류히하ㅓ나 예측하ㄴ

## 어떻게 특징( Feature )과 규칙( Rules ) 를 찾는가?
- 특징 추출 데이터가 어떤 특징을 가지고 있는지 찾고 백터로 만드는 처리과정
- 어떤 특징 ( Feature, Column )을 추출할지는 분석가가 결정한다. 머신러닝의 중요한 포인트
- 회귀분석( y=ax+b ) : 데이터가 연속된 값일때 사용하며, 이를 선형회귀 ( Linear )라고 부른다.<br>
  이때 y를 종속변수( Target )라고 칭한다. x를 독립변수( Feature ) 라고 칭한다. 
    - a와 b를 찾아내는 것이 머신러닝이다.
    - a = 기울기, b = 절편 / f(x) = ax + b : 그래프 그리던 1차함수
    - 과거 데이터를 이용해 (x), 미래 결과값을 예측 (y)하는 과정


## 머신러닝의 종류
#### 지도학습 ( Supervised Learning )
: 학습 시, 데이터와 답을 같이 학습한 후 다른 데이터의 답을 예측한다.     
#### 비지도학습 ( Unsupervised Learning )
: 학습 시, 데이터만 학습한 후 다른 데이터의 규칙성을 찾는다. <br>
- 클러스터 분석 ( Cluster Analysis ) - LLM 알고리즘의 시작점
- Clustering ( 분류 )
- *주성분 분석 ( Principal COmponent Analysis : PCA )    
#### 강화학습 ( Reinforcement Learning )
- 부분적으로 답을 입력한 후, 데이터를 기반으로 최적의 답을 찾아낸다.
    - 알파고 (AlphaGo)
    - 로봇
    - 게임 ( 방치형 등 )

## 머신러닝의 흐름
(1) 데이터 수집    
(2) 데이터 가공 - 결측치와 이상치를 처리하는 과정 -> 분석 (주로 산포도)    
(3) 데이터 학습    
(3-1) 학습 방법 선택 - 지도학습 / 비지도학습 등 | 판단 기준은 "예측력"    
(3-2) 매개변수( Hyper Parameter ) 조정    
(3-3) 모델 학습    
(4) 모델 평가
(5) 예측력( 정확도 )과 오분류표가 좋은 경우 다음 단계를 진행, 아니면 (3)부터 재실행    
(6) 성공

## 머신러닝의 응용 분야
#### 클래스 분류 ( Classification )
- 특정 데이터에 레이블( Target )을 분류할 수 있다.
- 스팸메일 분류, 필기인식, 증권사기
##### 이진분류
: 둘 중 하나의 값으로 분류하는 경우
##### 다중분류
: 여러 개의 분류값 중에서 하나의 값으로 예측하는 경우

#### 클러스터링 ( Clustering ) : 그룹 나누기
- 값의 유사성을 기반으로 데이터를 여러 그룹으로 나눌 수 있다.
- 사용자의 취향을 그룹으로 묶어 사용자 취향에 맞는 광고를 제공한다. ( 알고리즘 )

#### 추천 ( Recommendation )
- 특정 데이터를 기반으로 다른 데이터를 추천하는 것.
- 인터넷 서점에서 구매한 책들을 기반으로 다른 책을 추천

#### 차원 축소 ( Dimensionality Reduction )
- 데이터의 특성을 유지하면서 데이터의 양을 줄이는 것.
- 데이터를 시각화하거나, 구조를 추출해서 용량을 줄여 계산을 빠르게 하거나 메모리를 절약한다.
- 압축 프로그램
- 컬럼의 갯수에 따라 차원을 분류함. ( y=ax + b 의 x 의 갯수 = 차원 )

#### 머신러닝의 필수 개념
##### 과대적합 ( Overfitting )
- 학습데이터 외의 데이터에는 정확도( 예측력 )가 낮게 나오는 모델

##### 과소적합 ( Underfitting )
- 학습데이터보다 일반데이터가 정확도가 높게 나오는 모델
